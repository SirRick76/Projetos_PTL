```python
import requests
import json
import pandas as pd
#endpoint
url="https://archive-api.open-meteo.com/v1/archive"
 
#parametros
parametros={
    "latitude":41.57246,
    "longitude":-8.415021,
    "start_date":"2025-11-19",
    "end_date":"2025-11-23",
    "hourly":["temperature_2m","relative_humidity_2m","precipitation"],
    "timezone":"Europe/Lisbon"
}
 
# enviar o pedido
resposta=requests.get(url,params=parametros)
#converter resposta json para dicionario
dados=resposta.json()
 
print("Resposta da API:")
print(json.dumps(dados,indent=4,ensure_ascii=False))
# criação do dataFrame
# converter cololuna time para datetime
 
df["time"]=pd.to_datetime(df["time"])
 
df=pd.DataFrame(dados["hourly"])
# primeira linhas
df.head()
df.shape
#tipos de dados
df.info()
```

```python
#identificar o dia em que ocorreu a temperartura maxima
 
indice_dia_max=df["temperature_2m"].idxmax()
registo_maximo=df.loc[indice_dia_max]
print("Registo temperatura maxima: ",registo_maximo["time"])
print("Registo temperatura maxima: ",registo_maximo)
```

## Chave API

```python
import os 
from dotenv import load_dotenv

load_dotenv()

API_KEY= os.getenv("WEATHER_API_KEY")

if not API_KEY:
    raise ValueError("A chave API nao foi encontrada")
print ("Chave carregada com sucesso")
```

```python
import requests
import json

url="http://api.weatherapi.com/v1/current.json"

parametros = {
    "key": API_KEY,
    "q": "Braga, Portugal",
    "aqi": "no"
}
```

```python
resposta = requests.get(url, params=parametros)

resposta.raise_for_status()
dados= resposta.json()

print(json.dumps(dados,indent=4,ensure_ascii=False))
```
