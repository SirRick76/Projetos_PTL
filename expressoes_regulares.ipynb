```python
import re
import pandas as pd
```

Procura um único algarismo de 0 a 9

```python
texto = "O produto custa 25 euros"
resultado = re.findall(r"\d",texto)
resultado
```

Procura um ou mais algarismos consecutivos.

```python
texto2= "Encomenda 125, produto 450 e quantidade 20"
resultado2=re.findall(r"\d+",texto2)
resultado2
```

[^0-9] – Identificar tudo o que não seja um algarismo
Procura todos os caracteres que não sejam números entre 0 e 9.

```python
texto3= "Preco:€25.50"
resultado3 = re.findall(r"[^0-9]", texto3)
resultado3
```

Aplicação prática no Pandas

```python
dados = {
    "preco":["€25.50", "€18.90","€45.00"],
    "referencia":["PROD125","PROD450","PROD020"]
}

df= pd.DataFrame(dados)
```

1- Extrair os algarismos das referências

```python
df["numero"]=df["referencia"].str.extract(r"(\d+)")
```

2- Remover tudo o que não seja algarismo ou ponto dos precos

```python
df["preco"]=df["preco"].str.replace(r"[^0-9.]","",regex=True)
```

3- Converter os precos para valores numéricos

```python
df["preco"]=pd.to_numeric(df["preco"])
```
