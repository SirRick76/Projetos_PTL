Exercício 17 –Utilizando os DataFrames clientes e encomendas, crie uma tabela que apresente, para cada cliente: 
 Identificador do cliente; 
 Nome; 
 Quantidade de encomendas realizadas; 
 Valor total gasto; 
 Valor médio por encomenda. 
O resultado deve incluir apenas os clientes que realizaram pelo menos uma encomenda. 


```python
ex17=pd.merge(clientes,encomendas,on="cliente_id")
ex17=ex17.groupby(["cliente_id","nome"]).agg({
    "encomenda_id":"count",
    "valor":["sum","mean"]
}).reset_index()
ex17
```

Exercício 18 –Utilizando os DataFrames disponíveis, construa uma tabela final destinada a um sistema de gestão de 
clientes contendo: 
 ID do cliente; 
 Nome; 
 Cidade; 
 Email; 
 Telefone; 
 Número de encomendas; 
 Valor total gasto. 
Ordene os resultados por valor total gasto, do maior para o menor. 

```python
ex18=pd.concat([clientes,novos_clientes])
ex18=pd.merge(ex18,encomendas,on="cliente_id",how="left")
ex18=ex18.groupby(["cliente_id","nome"]).agg({
    "encomenda_id":"count",
    "valor":"sum"
}).sort_values(by=["valor","cliente_id"],ascending=[False,True]).reset_index()
ex18=pd.merge(ex18,contactos,on="cliente_id",how="left")
ex18
```
