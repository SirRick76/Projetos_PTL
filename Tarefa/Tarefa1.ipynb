## Representação gráfica dos dados

3- Crie um gráfico de barras que represente o número de ofertas de emprego por profissão (job_title_short).

```python
profissao=df["job_title_short"].value_counts()
plt.bar(profissao.index,profissao.values)
plt.title("Número de ofertas de emprego por profissao")
plt.xlabel("Profissao")
plt.ylabel("Número de ofertas")
plt.show()
```

4- Crie um histograma que permita observar a distribuição dos salários anuais (salary_year_avg).

```python
df["salary_year_avg"].hist()
plt.title("Distribuição dos salários anauis")
plt.xlabel("Valor salário anual (€)")
plt.ylabel("Frequência")
plt.show()
```
