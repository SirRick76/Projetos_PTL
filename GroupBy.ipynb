```python
import pandas as pd
```

```python
dados={
    "Comercial":["Ana","Pedro","Ana","Rui","Pedro", "Ana","Rui","Pedro"],
    "Região":["Norte","Sul","Norte","Centro","Sul","Centro","Norte","Centro"],
    "Produto":["Computador","Tablet","Tablet","Computador","Computador","Tablet","Computador","Tablet"],
    "Vendas":[1500,800,600,1200,1800,900,1100,700],
    "Unidades":[3,2,2,2,4,3,2,2]
}

df=pd.DataFrame(dados)
df.groupby("Produto")["Vendas"].sum()


```

```python
df.groupby("Produto")["Vendas"].mean()
```

```python
df.groupby("Produto")["Vendas"].max()
```

```python
df.groupby("Produto")["Vendas"].min()
```

```python
df.groupby("Produto").size()
```

```python
df.groupby("Produto")["Vendas"].agg(["sum","mean","min","max","count"])
```

```python
df.groupby("Comercial").agg({
    "Vendas":"sum",
    "Unidades":"mean",
    "Produto":"count"
})
```

```python
df.groupby(["Comercial","Região"])["Vendas"].sum()
```

```python
resultado= df.groupby("Comercial")["Vendas"].sum()

print(resultado.index)
print(resultado)
```

```python
df.groupby("Comercial", as_index=False)["Vendas"].sum()
```
