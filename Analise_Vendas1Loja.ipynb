```python
df["Produto"]=df["Produto"].fillna("Não Identificado")
df["Categoria"]=df["Categoria"].fillna("Não Identificada")
df["Cidade"]=df["Cidade"].fillna("Não Identificada")
df["Cliente"]=df["Cliente"].fillna("Não Identificado")
#df["Cidade"].value_counts()
#df["Produto"].value_counts()
#df["Categoria"].value_counts()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["Data_Venda"]=pd.to_datetime(df["Data_Venda"],format="%d/%m/%Y")
```

```python
df["ID_Venda"]=df["ID_Venda"].astype(str)
df["Quantidade"]=df["Quantidade"].astype(int)
df["Preco_Unitario"]=df["Preco_Unitario"].astype(float)
```

```python
df["Cidade"]=df["Cidade"].str.strip()
df["Cidade"]=df["Cidade"].str.title()
```

```python
df["Cidade"].value_counts()
```

```python
df["Cidade"]=df["Cidade"].replace("Setubal","Setúbal")
```

```python
df["Cidade"].value_counts()
```

```python
df=df.drop(columns=["Observacoes"])
```
