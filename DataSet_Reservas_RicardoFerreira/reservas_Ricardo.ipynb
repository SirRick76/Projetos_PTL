13- Formatação do tipo de dados

```python
df.dtypes
```

```python
df["Data_Entrada"]=pd.to_datetime(df["Data_Entrada"],format="%d/%m/%Y")
df["Data_Saida"]=pd.to_datetime(df["Data_Saida"],format="%d/%m/%Y")
```

```python
df["Numero_Hospedes"]=df["Numero_Hospedes"].astype("Int64")
df["Desconto_Percentagem"]=df["Desconto_Percentagem"].astype(float)
df["Avaliacao_Cliente"]=df["Avaliacao_Cliente"].astype(float)
```

```python
df["Pais"]=df["Pais"].str.title()
df["Tipo_Quarto"]=df["Tipo_Quarto"].str.title()
df["Canal_Reserva"]=df["Canal_Reserva"].str.title()
df["Estado_Reserva"]=df["Estado_Reserva"].str.title()
```

14- Tratamento de dados em falta

```python
df.isna().sum()
```

```python
mapa_paises = df.drop_duplicates("Cidade").set_index("Cidade")["Pais"]
df["Cidade"]=df["Cidade"].fillna("Não Identificada")
df["Pais"] = df["Pais"].fillna(df["Cidade"].map(mapa_paises))
df["Tipo_Quarto"]=df["Tipo_Quarto"].fillna("Não Identificado")
df["Canal_Reserva"]=df["Canal_Reserva"].fillna("Canal Não Identificado")
df["Estado_Reserva"]=df["Estado_Reserva"].fillna("Estado Não Identificado")
df["Data_Entrada"] = df["Data_Entrada"].fillna(
    df["Data_Saida"] - pd.to_timedelta(df["Numero_Noites"], unit="D"))
```

```python
df = df.dropna(subset=["Numero_Hospedes"])
df = df.dropna(subset=["Preco_Por_Noite"])
```

15-Eliminar colunas desnecessarias

```python
df = df.drop(columns=["Avaliacao_Cliente", "Observacoes"])
```

```python
df.info()
```

16-Criação de novas colunas e Transformação de Dados

Anos(Entrada e Saida)

```python
df["Ano_Entrada"]=df["Data_Entrada"].dt.year
df["Ano_Saida"]=df["Data_Saida"].dt.year
```

Mes(Entrada e Saida)

```python
df["Mes_Entrada"]=df["Data_Entrada"].dt.month
df["Mes_Saida"]=df["Data_Saida"].dt.month
```

Ano_Mes(Entrada e Saida)

```python
df["Ano_Mes_Entrada"]=df["Data_Entrada"].dt.strftime("%Y-%m")
df["Ano_Mes_Saida"]=df["Data_Entrada"].dt.strftime("%Y-%m")
```

Trimestre(Entrada e Saida)

```python
df["Trimestre_Entrada"]=df["Data_Entrada"].dt.quarter
df["Trimeste_Saida"]=df["Data_Saida"].dt.quarter
```

Dia da Semana(Entrada e Saida)

```python
df["Dia_Semana_Entrada"]=df["Data_Entrada"].dt.day_name(locale="PT")
df["Dia_Semana_Saida"]=df["Data_Saida"].dt.day_name(locale="PT")
```

Preco_Total_Sem_Desconto

```python
df["Preco_Total_Sem_Desconto"]=(df["Preco_Por_Noite"]* df["Numero_Noites"])
```

Valor_Desconto

```python
df["Valor_Desconto"]=(df["Preco_Total_Sem_Desconto"]*(df["Desconto_Percentagem"]/100))
```

Classificacao do tipo de dia da entrada da reserva(Fim de semana ou semana) 

```python
df["TipoDiadaSemanaEntrada"]= ""

#Classificar as vendas
df.loc[(df["Dia_Semana_Entrada"] == "Sábado") | (df["Dia_Semana_Entrada"]) == "Domingo", "TipoDiadaSemanaEntrada"] = "Fim de Semana"
df.loc[(df["Dia_Semana_Entrada"] != "Sábado") & (df["Dia_Semana_Entrada"] != "Domingo"), "TipoDiadaSemanaEntrada"] = "Semana"
```

```python
df[["Dia_Semana_Entrada","TipoDiadaSemanaEntrada"]]
```

Categoria Preco(Economico,medio ou premium)

```python
def classificar_quarto(preco):
    if preco<100:
        return "Economico"
    elif preco>=100  and preco<180:
        return "Médio"
    else:
        return "Premium"
```

```python
df["Classificacao_Quarto"] = df["Preco_Por_Noite"].apply(classificar_quarto)
```
