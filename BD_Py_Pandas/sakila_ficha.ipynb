# Fase 1 — Ligação à base de dados e importação de dados 

1-Instale as bibliotecas necessárias para trabalhar com MySQL e Pandas

```python
from sqlalchemy import create_engine
import pandas as pd
from dotenv import load_dotenv
import os
```

2-Configure a ligação à base de dados utilizando as credenciais guardadas num ficheiro .env. 
3- Crie o engine de ligação através do SQLAlchemy. 
4- Teste a ligação à base de dados. 

```python
load_dotenv()
```

```python
utilizador=os.getenv("DB_USER")
password=os.getenv("DB_PASSWORD")
servidor=os.getenv("DB_HOST")
porta = int(os.getenv("DB_PORT"))
base_dados=os.getenv("DB_NAME")

engine=create_engine(f"mysql+pymysql://{utilizador}:{password}@{servidor}:{porta}/{base_dados}")
with engine.connect() as conexao :
    print("Ligação estabelecida com sucesso!") 
```

5- Importe a tabela film para um DataFrame. 

```python
def importar_dados(consulta):
    with engine.connect() as conexao:
        df = pd.read_sql(consulta, conexao)
    return df
```

```python
consulta_filmes="SELECT * FROM film"
df_filmes=importar_dados(consulta_filmes)
df_filmes
```

6- Importe as tabelas category, customer, rental e payment para DataFrames independentes. 

```python
consulta_categorias="SELECT * FROM category"
df_categorias=importar_dados(consulta_categorias)
df_categorias
```

```python
consulta_aluguer="SELECT * FROM rental"
df_aluguer=importar_dados(consulta_aluguer)
df_aluguer
```

```python
consulta_pagamento="SELECT * FROM payment"
df_pagamentos=importar_dados(consulta_pagamento)
df_pagamentos
```

4-Identifique a categoria com maior número de filmes.

```python
df_filmes_cat_qtd[df_filmes_cat_qtd==df_filmes_cat_qtd.min()]
```

6-Calcule a duração média dos filmes por categoria. 

```python
df_filmes_cat1=pd.merge(df_filmes,df_film_category,on="film_id")
df_filmes_cat1=pd.merge(df_filmes_cat1,df_categorias,on="category_id")
df_filmes_cat1 = df_filmes_cat1[["title","name","length"]]
df_filmes_cat1 = df_filmes_cat1.rename(columns={"name":"categoria"})
df_med_fil_cat= df_filmes_cat1.groupby("categoria")["length"].mean()
df_med_fil_cat
```
