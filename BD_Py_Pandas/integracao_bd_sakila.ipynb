```python
import os
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine,text
```

```python
load_dotenv()
```

```python
utilizador=os.getenv("DB_USER")
password=os.getenv("DB_PASSWORD")
servidor=os.getenv("DB_HOST")
porta = int(os.getenv("DB_PORT"))
engine=create_engine(f"mysql+pymysql://{utilizador}:{password}@{servidor}:{porta}")

#criar a bd
sql = "CREATE DATABASE IF NOT EXISTS vendas_informatica"

with engine.connect() as conexao:
    conexao.execute(text(sql))
    conexao.commit
 
```

```python
dados = {
    "Comercial": ["Ana", "Pedro", "Ana", "Rui",
                  "Pedro", "Ana", "Rui", "Pedro"],
 
    "Região": ["Norte", "Sul", "Norte", "Centro",
               "Sul", "Centro", "Norte", "Centro"],
 
    "Produto": ["Computador", "Tablet", "Tablet", "Computador",
                "Computador", "Tablet", "Computador", "Tablet"],
 
    "Vendas": [1500, 800, 600, 1200, 1800, 900, 1100, 700],
 
    "Unidades": [3, 2, 2, 2, 4, 3, 2, 2]
}
df=pd.DataFrame(dados)

#
engine=create_engine(f"mysql+pymysql://{utilizador}:{password}@{servidor}:{porta}/vendas_informatica")

#criar a tabela e inserir os dados
df.to_sql(name="venda",con=engine,if_exists="replace",index=False)
```
