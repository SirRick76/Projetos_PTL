## Módulo 1 – Criação de variáveis temporais

**2.** Crie uma coluna denominada Mes_Publicacao, contendo o número correspondente ao mês de publicação.

**3.** Crie uma coluna denominada Dia_Semana, contendo o nome do dia da semana em que a oferta foi publicada.

## Módulo 2 – Criação e transformação de variáveis

**6.** Crie uma nova coluna denominada Tipo_Trabalho, a partir da coluna job_work_from_home (True → Remoto; False → Presencial).

```python
# criar uma coluna vazia
df["Tipo_Trabalho"] = ""

# classificar as ofertas
df.loc[df["job_work_from_home"] == True, "Tipo_Trabalho"] = "Remoto"
df.loc[df["job_work_from_home"] == False, "Tipo_Trabalho"] = "Presencial"

df["Tipo_Trabalho"].value_counts()
```

**7.** Crie uma coluna denominada Nivel_Salarial, utilizando a informação da coluna salary_year_avg. Os valores em falta devem continuar sem classificação.

```python
# a coluna começa com valores em falta (None), para que as ofertas
# sem salário continuem sem classificação, como pede o enunciado
df["Nivel_Salarial"] = None

df.loc[df["salary_year_avg"] < 40000, "Nivel_Salarial"] = "Baixo"
df.loc[(df["salary_year_avg"] >= 40000) & (df["salary_year_avg"] < 70000), "Nivel_Salarial"] = "Médio"
df.loc[(df["salary_year_avg"] >= 70000) & (df["salary_year_avg"] < 100000), "Nivel_Salarial"] = "Elevado"
df.loc[df["salary_year_avg"] >= 100000, "Nivel_Salarial"] = "Muito elevado"

df["Nivel_Salarial"].value_counts(dropna=False)
```

```python
# confirmação: as ofertas sem classificação são exatamente as que não têm salário anual
print(df["Nivel_Salarial"].isna().sum(), df["salary_year_avg"].isna().sum())
```

As comparações com NaN (`<`, `>=`) dão sempre `False`, por isso as ofertas sem salário não entram em nenhum `loc[]` e mantêm o valor em falta inicial.

**8.** Crie uma coluna denominada Tipo_Oferta: "Salário divulgado" quando existe valor em salary_year_avg ou em salary_hour_avg; "Salário anual não divulgado" quando ambas as colunas não têm valor.

```python
df["Tipo_Oferta"] = "Salário anual não divulgado"
df.loc[(df["salary_year_avg"].notna()) | (df["salary_hour_avg"].notna()), "Tipo_Oferta"] = "Salário divulgado"

df["Tipo_Oferta"].value_counts()
```

## Módulo 3 – Aplicação de funções com apply()

**9.** Crie uma função denominada classificar_salario() que receba um valor salarial e devolva a classificação correspondente.

```python
def classificar_salario(salario):
    if pd.isna(salario):          # sem salário → sem classificação
        return None
    elif salario < 40000:
        return "Baixo"
    elif salario < 70000:         # já sabemos que é >= 40000
        return "Médio"
    elif salario < 100000:        # já sabemos que é >= 70000
        return "Elevado"
    else:
        return "Muito elevado"

# testes, incluindo os valores-limite dos escalões
for valor in [35000, 40000, 69999, 70000, 100000, np.nan]:
    print(valor, "→", classificar_salario(valor))
```

A verificação `pd.isna()` vem primeiro porque um NaN falha todas as comparações e, sem ela, cairia no `else` ("Muito elevado").

**10.** Utilize o método apply() para aplicar a função à coluna salary_year_avg, guardando os resultados numa nova coluna denominada Classificacao_Salario.

```python
df["Classificacao_Salario"] = df["salary_year_avg"].apply(classificar_salario)
df["Classificacao_Salario"].value_counts(dropna=False)
```

**11.** Compare os resultados obtidos com os da coluna Nivel_Salarial, criada anteriormente.

```python
# nº de linhas com classificações diferentes
# (fillna só para a comparação, não altera o DataFrame)
(df["Nivel_Salarial"].fillna("Sem salário") != df["Classificacao_Salario"].fillna("Sem salário")).sum()
```

```python
pd.crosstab(df["Nivel_Salarial"], df["Classificacao_Salario"])
```

As duas colunas são **iguais em todas as linhas**: a tabela cruzada só tem valores na diagonal. Os dois métodos (`loc[]` com condições e `apply()` com uma função) chegam ao mesmo resultado. O `loc[]` é mais direto para regras simples; a função é mais fácil de reutilizar e de testar.
