9-Aplique o tratamento escolhido:
Substituição de valores em falta, quando justificado.
Eliminação de registos, quando a ausência de informação comprometer a sua utilização.

```python
df["company_name"]=df["company_name"].fillna("Não Identificado")
df["job_title"]=df["job_title"].fillna(df["job_title_short"])
df["job_via"]=df["job_via"].fillna("Via Não Identificada")
df["job_country"]=df["job_country"].fillna(df["job_location"])
df["job_location"]=df["job_location"].fillna(df["job_country"])

```
