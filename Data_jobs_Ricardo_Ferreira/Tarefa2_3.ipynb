## Módulo 1 – Criação de variáveis temporais

1-Crie uma coluna denominada Ano_Publicacao, contendo o ano de publicação da oferta.

```python
df["Ano_Publicacao"]=df["job_posted_date"].dt.year
```

2-Crie uma coluna denominada Mes_Publicacao, contendo o número correspondente ao mês de publicação.

```python
df["Mes_Publicacao"]=df["job_posted_date"].dt.month
```

3-Crie uma coluna denominada Dia_Semana, contendo o nome do dia da semana em que a oferta foi publicada.

```python
df["Dia_Semana"]=df["job_posted_date"].dt.day_name(locale="PT")
```

4-Crie uma coluna denominada Trimestre, identificando o trimestre correspondente à data de publicação.

```python
df["Trimestre"]=df["job_posted_date"].dt.quarter
```

5-Crie uma coluna denominada Ano_Mes, apresentando o ano e o mês no formato AAAA-MM.

```python
df["Ano_Mes"]=df["job_posted_date"].dt.strftime("%Y-%m")
```

## Módulo 2 – Criação e transformação de variáveis

6-Crie uma nova coluna denominada Tipo_Trabalho, a partir da coluna job_work_from_home.

```python
#criar uma coluna vazia
df["Tipo_Trabalho"]= ""

#Classificar as vendas
df.loc[df["job_work_from_home"] == True, "Tipo_Trabalho"] = "Remoto"
df.loc[df["job_work_from_home"] == False, "Tipo_Trabalho"] = "Presencial"
```

```python
df.head(5)
```

7-Crie uma coluna denominada Nivel_Salarial, utilizando a informação da coluna salary_year_avg.

```python
#criar uma coluna vazia
df["Nivel_Salarial"] = "Sem informação"

df.loc[df["salary_year_avg"].notna() & (df["salary_year_avg"] < 40000), "Nivel_Salarial"] = "Baixo"

df.loc[(df["salary_year_avg"] >= 40000) & (df["salary_year_avg"] < 70000), "Nivel_Salarial"] = "Médio"

df.loc[(df["salary_year_avg"] >= 70000) & (df["salary_year_avg"] < 100000), "Nivel_Salarial"] = "Elevado"

df.loc[df["salary_year_avg"] >= 100000, "Nivel_Salarial"] = "Muito elevado"
```

```python
df[["Nivel_Salarial", "salary_year_avg"]].head(10)
```

8-Crie uma coluna denominada Tipo_Oferta, que permita identificar se a oferta apresenta ou não informação sobre o salário.

```python
df["Tipo_Oferta"] = ""

df.loc[(df["salary_year_avg"].isna()) & (df["salary_hour_avg"].isna()),"Tipo_Oferta"] = "Salário não divulgado"
df.loc[(df["salary_year_avg"].notna()) | (df["salary_hour_avg"].notna()),"Tipo_Oferta"] = "Salário divulgado"
```

```python
df[["salary_year_avg", "salary_hour_avg", "Tipo_Oferta"]].head(50)
```

## Módulo 3 – Aplicação de funções com apply()

9-Crie uma função denominada classificar_salario() que receba um valor salarial e devolva a classificação correspondente

```python
def classificar_salario(salario):
    if pd.isna(salario):
        return "Sem informação"
    elif salario<40000:
        return "Baixo"
    elif salario>40000 and salario<70000:
        return "Médio"
    elif salario>70000 and salario<100000:
        return "Elevado"
    else:
        return "Muito elevado"
```

10-Utilize o método apply() para aplicar a função à coluna salary_year_avg, guardando os resultados numa nova coluna denominada Classificacao_Salario.

```python
df["Classificacao_Salario"] = df["salary_year_avg"].apply(classificar_salario)
```
