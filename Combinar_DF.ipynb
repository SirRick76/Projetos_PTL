```python
import pandas as pd
```

# Merge

```python
vendas=pd.DataFrame({
    "ID_Venda" : [1,2,3,4],
    "ID_Produto": [101,102,103,101],
    "Quantidade": [2,3,1,5]
})
vendas
```

```python
produtos=pd.DataFrame({
    "ID_Produto": [101,102,103],
    "Produto": ["Computador","Monitor", "Teclado"],
    "Preco":[800,150,30]
})
produtos
```

```python
resultado= pd.merge(vendas,produtos,on="ID_Produto")
resultado
```

```python
vendas=pd.DataFrame({
    "ID_Venda" : [1,2,3,4],
    "ID_Produto": [101,102,103,104],
    "Quantidade": [2,3,1,5]
})
vendas
```

```python
produtos=pd.DataFrame({
    "ID_Produto": [101,102,103,105],
    "Produto": ["Computador","Monitor", "Teclado","Rato"],
    "Preco":[800,150,30,20]
})
produtos
```

```python
resultado= pd.merge(vendas,produtos,how="inner")
resultado
```

```python
resultado= pd.merge(vendas,produtos,how="left")
resultado
```

```python
resultado= pd.merge(vendas,produtos,how="right")
resultado
```

```python
resultado= pd.merge(vendas,produtos,how="outer")
resultado
```

```python
vendas= pd.DataFrame({
    "ID_Produto":[101,102,103],
    "Quantidade":[2,3,1]
})
vendas
```

# Concat

```python
vendas_janeiro=pd.DataFrame({
    "Produto":["Computador", "Monitor"],
    "Quantidade":[10,20]
})
vendas_janeiro
```

```python
vendas_fevereiro=pd.DataFrame({
    "Produto":["Teclado", "Rato"],
    "Quantidade":[30,40]
})
vendas_fevereiro
```

```python
resultado=pd.concat([vendas_janeiro,vendas_fevereiro])
resultado
```

```python
resultado=pd.concat([vendas_janeiro,vendas_fevereiro],ignore_index=True)
resultado
```

```python
produtos=pd.DataFrame({
    "ID":[101,102,103],
    "Produto":["Computador","Monitor","Teclado"]
})
produtos
```

```python
precos= pd.DataFrame({
    "Preco":[800,150,30],
    "Stock":[10,25,50]
})
precos
```

```python
resultado=pd.concat([produtos,precos],axis=1)
resultado
```

# Cuidados importantes

```python
#Identificadores repetidos: ter cuidado com o merge

vendas= pd.DataFrame({
    "ID_Produto":[101,101,103],
    "Quantidade":[2,3,1]
})
produtos=pd.DataFrame({
    "ID_Produto":[101,101,103],
    "Produto":["Computador","Monitor","Teclado"]
})
produtos
resultado=pd.merge(vendas,produtos,on="ID_Produto")
resultado
```

```python
print(produtos["ID_Produto"].unique())
print(produtos["ID_Produto"].is_unique)
```

```python
resultado=pd.merge(vendas,produtos,on="ID_Produto", validate="many_to_one")
```

```python
resultado=pd.merge(vendas,produtos,on="ID_Produto", validate="many_to_many")
```

```python
# parametro indicador
#recebemos 2 tabelas e queremos saber se todos os registos tem correspondencia
resultado=pd.merge(vendas,produtos,on="ID_Produto",how="outer",indicator=True)
resultado
```

```python
vendas=pd.DataFrame({
    "ID_Venda": [1,2,3],
    "ID_Produto":[101,102,103],
    "Quantidade":[2,3,1],
    "ID_Comercial":[1,2,1]
})
 
produtos=pd.DataFrame({
    "ID_Produto": [101, 102, 103],
    "Produto": ["Computador", "Monitor", "Teclado"],
    "Preco": [800, 150, 30]
})
comercial=pd.DataFrame({
    "ID_Comercial":[1,2],
    "Comercial":["Ana","João"]
})
```

```python
resultado=pd.merge(vendas,produtos,on="ID_Produto",how="inner")
resultado
```

```python
resultado2=pd.merge(resultado,comercial,on="ID_Comercial",how="inner")
resultado2
```
