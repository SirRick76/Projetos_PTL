```python
import pandas as pd
# 1. Criar o mini dataset de vendas
dados = {
 
    'Produto': ['Smartphone', 'Camisa', 'Cafeteira', 'Laptop', 'Calça', 'Liquidificador', 'Fone de Ouvido', 'Casaco', 'Teclado', 'Tênis'],
    'Categoria': ['Eletrónicos', 'Vestuário', 'Eletrodomésticos', 'Eletrónicos', 'Vestuário', 'Eletrodomésticos', 'Eletrónicos', 'Vestuário', 'Eletrónicos', 'Vestuário'],
    'Valor (€)': [599, 29, 49, 899, 39, 35, 79, 89, 45, 69]
}
 
df1= pd.DataFrame(dados)
```

```python
df1.head(10)
```

```python
df1["Categoria"].value_counts().plot(
    kind="bar",
    title="Nº de produtos",
    xlabel="Categoria",
    ylabel="Nº de Produtos"
)
```

```python
df1["Categoria"].value_counts().plot(
    kind="barh",
    title="Nº de produtos",
    xlabel="Categoria",
    ylabel="Nº de Produtos"
)
```

```python
df1["Categoria"].value_counts(ascending=True).plot(
    kind="barh",
    title="Nº de produtos",
    xlabel="Categoria",
    ylabel="Nº de Produtos"
)
```

```python
df1["Categoria"].value_counts().plot(
    kind="pie",
    title="Distribuição de produto por categoria",
    autopct='%.0f%%',
    legend=True
    
)
```

Metodos .loc[]e .iloc[]

```python
import pandas as pd
```

```python
dados={
    "Produto":["Computador","Monitor","Teclado","Rato"],
    "Categoria":["Informatica","Acessórios","Informatica","Acessórios"],
    "Preco":[750,180,35,20]
}
# criar o dataframe
 
df=pd.DataFrame(dados)
df
```

diferença esta na forma como identificam linhas e colunas no dataframe
loc[]-> seleciona atraves de rotulos(quando temos nomes dos indices das colunas)
iloc[]-> seleciona atraves de posiçoes numericas (posicao de linhas e colunas) 

```python
#a sintaxe do loc[linha,coluna]
#seleciona 1 linha
df.loc[1]
```

```python
#selecionar uma celula
df.loc[0,"Categoria"]
```

```python
#selecionar varias colunas
df.loc[0,["Produto","Preco"]]
```

```python
df.loc[1:2,["Produto","Preco"]]
```

```python
df.loc[df["Preco"]>=35]
```

```python
#alterar a categoria dos produtos cujo preco e >=500 para produto premium 
df.loc[df["Preco"]>=500,"Categoria"]="Produto premium"
df
```

```python
#seleciona os dados atraves de posiçoes numericas
#
#seleciona um valor especifico
df.iloc[1,1]
```

```python
#seleciona varias linha
df.iloc[0:3] # linhas posiçoes 0,1,2
```

```python
#seleciona varias colunas
df.iloc[:,[0,1]]
```
